# 01 - Data Exploration
Explore the pre-computed Doppler traces (no raw CFR here -- see project constraint #2).

```python
import sys
sys.path.insert(0, '..')

from src.data.doppler_trace_dataset import DopplerTraceDataset, discover_stream_files
from src.data.label_mapping import TARGET_CLASSES

root = '../data/doppler_traces'
files = discover_stream_files(root)
print(f'Discovered {len(files)} stream files')
```

```python
ds = DopplerTraceDataset(root, sets_to_include=('S1',))
print(ds.summary())
print('Target classes:', TARGET_CLASSES)
```

```python
import matplotlib.pyplot as plt

x, y = ds[0]
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ant in range(4):
    axes[ant].imshow(x[ant].numpy().T, aspect='auto', origin='lower', cmap='viridis')
    axes[ant].set_title(f'Antenna {ant}')
fig.suptitle(f'Sample 0 -- class index {y}')
plt.show()
```
